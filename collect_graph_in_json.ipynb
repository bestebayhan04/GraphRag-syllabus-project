```python
from json import load

with open("data/courses/scripts.json", "r") as f:
    courses = load(f)

nodes = {"courses": [], "occupations": [], "skills": []}
edges = []
for course in courses:
    n_c = {
        "course code": course["course_code"],
        "course title": course["course_title"],
        "source file": course["source_file"],
        "topics": course["topics"],
        "learning outcomes": course["learning_outcomes"]
    }

    if course["course_code"] == "":
        print(course)
    nodes["courses"].append(n_c)

    for pre in course["prerequisites"]:
        edges.append([pre, course["course_code"], "PREREQ"])
```

```python
from glob import glob

files = glob("./data/careers/*.json")
skill_ids = set()

for file in files:
    with open(file, "r") as f:
        occupation = load(f)

    n_o = {
        "id": occupation["id"],
        "title": occupation["mainLabel"],
        "alternative titles": occupation["alternativeLabels"],
        "description": occupation["description"]
    }

    nodes["occupations"].append(n_o)

    for skill in occupation["skills"]:
        if skill["id"] not in skill_ids:
            n_s = {
                "id": skill["id"],
                "title": skill["title"]
            }

            nodes["skills"].append(n_s)
            edges.append([occupation["id"], skill["id"], "REQUIRES"])
```

```python
with open("outputs/triplets/triples.json", "r") as f:
    triples = load(f)

for triple in triples:
    if triple["relation"] == "COVERS":
        edges.append([triple["subject"], triple["object"], "COVERS"])
```

```python
from json import dump

graph_knowledge = {"nodes": nodes, "edges": edges}

with open("graph_database/database.json", "w") as f:
    dump(graph_knowledge, f)
```
